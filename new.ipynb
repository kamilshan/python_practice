Step 1 Key preperation 

```python
#str.maketrans() IT creates a mapping of characters to be replaced in a string. It takes two arguments: the first is a string of characters to be replaced, and the second is a string of characters to replace them with. The two strings must be of equal length. The function returns a translation table that can be used with the str.translate() method to perform the replacements.
#str.translate() IT takes a translation table as an argument and returns a new string with the characters replaced according to the mapping defined in the translation table. It is used to perform character replacements in a string based on the mapping created by str.maketrans().
abc="ABCDEFGHIJKLMNOPQRSTUVWXYZ"


key=abc[3:]+abc[:3]
print(key)

#step 2: create a secret aplphabet by shifiting 3 positions 
#  step 3 our secret message 
msg=" MEET ME AT CANTEEN"
#STEP 4 CREATE THE ENCRYPTION MAPPING
emap= str.maketrans(abc,key)
#step 5: encrypt the message
cipher_text=msg.translate(emap)
print("SENDER")
print("Original Message:",msg)
print("Encrypted Message:",cipher_text)
#step 6: REciever - create decryption mapping 
dmap= str.maketrans(key,abc)
#step 7: decrypt the message
original_message=cipher_text.translate(dmap)
print("\n Reciever")
print("received message:",cipher_text)
print("decrypted message:",original_message)
```
