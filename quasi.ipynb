Slicing

#slicing means extracting a portion of a string. The syntax for slicing is s[start:end], where start is the index of the first character to include, and end is the index of the first character to exclude. For example, s[1:4] will give you the substring "HAT" from "CHATGPT".
# start is included and end is excluded. If start is omitted, it defaults to 0. If end is omitted, it defaults to the length of the string. You can also use negative indices to slice from the end of the string. For example, s[-3:-1] will give you the substring "GP" from "CHATGPT".

```python
#slicing means extracting a portion of a string. The syntax for slicing is s[start:end], where start is the index of the first character to include, and end is the index of the first character to exclude. For example, s[1:4] will give you the substring "HAT" from "CHATGPT".
# start is included and end is excluded. If start is omitted, it defaults to 0. If end is omitted, it defaults to the length of the string. You can also use negative indices to slice from the end of the string. For example, s[-3:-1] will give you the substring "GP" from "CHATGPT".
```

```python
s="hello"
print(s[1:4],s[1:],s[:4],s[:],s[-3:-1],s[-3:],s[:-1],s[:-3],s[-1:])
# by default  python increments the index by +1 
```

```python
print(s[2:200]) # if the end index is greater than the length of the string, it will return the substring from start index to the end of the string. It will not give an error.
```

```python
#%% . 5 the step :  s[start:end:step]  # step is the increment value for the index. It can be positive or negative. If step is omitted, it defaults to 1. If step is negative, the string is sliced in reverse order.
s="python"
print(s[0:6:2],s[0:6:3],s[0:6:4],s[0:6:5],s[0:6:6])
print(s[0:6:-1],s[5:0:-1],s[5:0:-2],s[5:0:-3],s[5:0:-4],s[5:0:-5])
```

A negative step tells python to move from  right to left 

```python
value ="malayalam"
print(value[0:9:1]==value[9::-1]) # this will check if the string is a palindrome or not. A palindrome is a string that reads the same forwards and backwards. The first slice value[0:9:1] gives the original string, and the second slice value[9::-1] gives the reversed string. If they are equal, then the string is a palindrome.)
```
